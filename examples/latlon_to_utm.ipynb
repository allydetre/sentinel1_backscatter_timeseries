## Convert Lat/Lon Coordinates to UTM

```python
# import packages
import pandas as pd
import geopandas as gpd
from pyproj import Proj
```

### Load SNOTEL station data with lat/lon

```python
# df of SNOTEL coordinates
snotelcoords = pd.read_csv("../import/df/testsnotel.csv") # 6 site test df
```

### Get easting/northing coordinates for both exact and non-exact lat/lon SNOTEL coordinates

```python
# Take in lat/lon, convert to northing/easting and add epsg column:
snotelcoords['easting_exact'] = None
snotelcoords['northing_exact'] = None
snotelcoords['easting'] = None
snotelcoords['northing'] = None
snotelcoords['epsg'] = None

# Loop through each row in df
for index, row in snotelcoords.iterrows():
    lon_exact = row['lon_exact']
    lat_exact = row['lat_exact']
    
    # Determine the UTM zone and EPSG code based on longitude
    if -126 <= lon_exact < -120:
        utm_zone = 10
        epsg = 32610
    elif -120 <= lon_exact < -114:
        utm_zone = 11
        epsg = 32611
    elif -114 <= lon_exact < -108:
        utm_zone = 12
        epsg = 32612
    elif -108 <= lon_exact < -102:
        utm_zone = 13
        epsg = 32613
    elif -102 <= lon_exact < -96:
        utm_zone = 14
        epsg = 32614
    else:
        print(f"Longitude {lon_exact} is out of bounds for UTM zones in the western US.")
        continue  # Skip row if the longitude is out of bounds - can add more zones to this as needed!

    # Define the UTM projection
    utmproj = Proj(proj='utm', zone=utm_zone, ellps='WGS84')

    # Transform the coordinates
    easting_exact, northing_exact = utmproj(lon_exact, lat_exact)
    
    # Assign the transformed coords to dataframe
    snotelcoords.at[index, 'easting_exact'] = easting_exact
    snotelcoords.at[index, 'northing_exact'] = northing_exact
    snotelcoords.at[index, 'epsg'] = epsg

### again for the non exact coords
for index, row in snotelcoords.iterrows():
    lon = row['lon']
    lat = row['lat']
    
    # Determine the UTM zone and EPSG code based on longitude
    if -126 <= lon < -120:
        utm_zone = 10
        epsg = 32610
    elif -120 <= lon < -114:
        utm_zone = 11
        epsg = 32611
    elif -114 <= lon < -108:
        utm_zone = 12
        epsg = 32612
    elif -108 <= lon < -102:
        utm_zone = 13
        epsg = 32613
    elif -102 <= lon < -96:
        utm_zone = 14
        epsg = 32614
    else:
        print(f"Longitude {lon} is out of bounds for user-defined UTM zones in the western US.")
        continue  # Skip row if the longitude is out of bounds - can add more zones as needed!

    # Define UTM projection
    utmproj = Proj(proj='utm', zone=utm_zone, ellps='WGS84')

    # Transform coordinates
    easting, northing = utmproj(lon, lat)
    
    # Assign transformed coords to df
    snotelcoords.at[index, 'easting'] = easting
    snotelcoords.at[index, 'northing'] = northing


# Save to csv
snotelcoords.to_csv("../output/df/testsnotel_utm.csv")
```
